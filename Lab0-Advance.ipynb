Code 1 numpy exercises

```python
import numpy as np
my_mat = [[1,2,3],[4,5,6],[7,8,9]]
mat = np.array(my_mat)
print("Type/Class of this object:",type(mat))
print("Here is the matrix\n----------\n",mat,"\n----------")
print("Dimension of this matrix: ",mat.ndim,sep='') 
#ndim gives the dimensison, 2 for a matrix, 1 for a vector
print("Size of this matrix: ", mat.size,sep='') #size gives the total number of elements
print("Shape of this matrix: ", mat.shape,sep='') 
#shape gives the number of elements along each axes (dimension)
print("Data type of this matrix: ", mat.dtype,sep='') 
#dtype gives the data type contained in the array
```

```python
my_mat = [[1.1,2,3],[4,5.2,6],[7,8.3,9]]
mat = np.array(my_mat)
print("Data type of the modified matrix: ", mat.dtype,sep='') 
print("\n\nEven tuples can be converted to ndarrays...")
```

```python
b = np.array([(1.5,2,3), (4,5,6)])
print("We write b = np.array([(1.5,2,3), (4,5,6)])")
print("Matrix made from tuples, not lists\n---------------------------------------")
print(b)
```

```python
print("A series of numbers:",np.arange(5,16)) # A series of numbers from low to high
print("21 linearly spaced numbers between 1 and 5\n--------------------------------------------")
print(np.linspace(1,5,21))
```

```python
print("Vector of zeroes\n---------------------")
print(np.zeros(5))
```

```python
print("Matrix of zeroes\n--------------------")
print(np.zeros((3,4))) # Notice Tuples
```

```python
print("Vector of ones\n---------------------")
print(np.ones(5))
```

```python
print("Matrix of ones\n---------------------")
print(np.ones((5,2))) # Note matrix dimension specified by Tuples
```

```python
print("Matrix of 5's\n---------------------")
print(5*np.ones((3,5)))
```

```python
print("Empty matrix\n-------------\n", np.empty((3,5)))
```

```python
mat1 = np.eye(4)
print("Identity matrix of dimension", mat1.shape)
print(mat1)
```

Code 2 : Matplotlib exercise

```python
import matplotlib.pyplot as plt
people = ['Ann','Brandon','Chen','David','Emily','Farook',
'Gagan','Hamish','Imran','Julio','Katherine','Lily']
age = [21,12,32,45,37,18,28,52,5,40,48,15]
weight = [55,35,77,68,70,60,72,69,18,65,82,48]
height = [160,135,170,165,173,168,175,159,105,171,155,158]
```

```python
plt.scatter(age, height)
plt.show()
```

```python

plt.figure(figsize=(12,4))
plt.title("People's weight in kgs",fontsize=16, fontstyle='italic')
# Main plot function 'bar'
plt.bar(x=people,height=weight, width=0.6,color='orange',edgecolor='k',alpha=0.6)
plt.xlabel("People",fontsize=15)
plt.xticks(fontsize=14,rotation=30)
plt.yticks(fontsize=14)
plt.ylabel("Weight (in kgs)",fontsize=15)
plt.show()
```

```python
import numpy as np
plt.figure(figsize=(7,5))
# Main plot function 'hist'
plt.hist(weight,color='red',edgecolor='k', alpha=0.75,bins=5)
plt.title("Histogram of patient weight",fontsize=18)
plt.xlabel("Weight in kgs",fontsize=15)
plt.xticks(fontsize=15)
plt.yticks(fontsize=15)
plt.show()
```

Code 3 : Pandas exercise

```python
import numpy as np
import pandas as pd
# Create dataframe
data = {'Company':['GOOG','GOOG','MSFT','MSFT','FB','FB'],'Person':['Sam','Charlie','Amy','Vanessa','Carl','Sarah'],
'Sales':[200,120,340,124,243,350]}
df = pd.DataFrame(data)
df
```

```python
print("\nSum of the column 'Sales' is: ",df['Sales'].sum())
print("Mean of the column 'Sales' is: ",df['Sales'].mean())
print("Std dev of the column 'Sales' is: ",df['Sales'].std())
print("Min and max of the column 'Sales' are: ",df['Sales'].min(),"and",df['Sales'].max())
```

```python
print("\nMethod head() is for showing first few entries\n",'-'*50, sep='')
df.head()
```

```python
print("\nFinding unique values in 'col2'\n",'-'*40, sep='')
print(df['Person'].unique())
```
